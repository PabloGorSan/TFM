```python
column_to_number = {
    'a': 1,
    'b': 2,
    'c': 3,
    'd': 4,
    'e': 5,
    'f': 6,
    'g': 7,
    'h': 8
}
```

```python
def get_one_hot_move(move):
    one_hot_piece = np.zeros(64)
    one_hot_piece[((column_to_number.get(move[0]) - 1)) +  ((8 - int(move[1])) * 8)] = 1
    return one_hot_piece
```

```python
one_hot_move = get_one_hot_move([*move[:2]])
one_hot_move.shape = (8,8)
one_hot_move
```

```python
def get_initial_positions(move):
    return get_one_hot_move([*move[:2]])

def get_final_positions(move):
    return get_one_hot_move([*move[2:4]])
```

```python
aux_move = 'h7b8q'
aux_one = get_initial_positions(aux_move)
aux_one.shape = (8,8)
aux_one
```

```python
data['initial_pos'] = data['move'].apply(get_initial_positions)
data['final_pos'] = data['move'].apply(get_final_positions)
```
