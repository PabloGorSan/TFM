```python
import sqlite3
def create_sqlite_database(filename):
    """ create a database connection to an SQLite database """
    conn = None
    try:
        conn = sqlite3.connect(filename)
        print(sqlite3.sqlite_version)
    except sqlite3.Error as e:
        print(e)
    finally:
        if conn:
            conn.close()
```

```python
SQLITE_PATH = "./data/database2.db"
```

```python
create_sqlite_database(SQLITE_PATH)
```

```python
def add_csv_data(df, db):
    conn = sqlite3.connect(db)
    df.to_sql("moves", conn, if_exists='append', index=False)
    conn.close()
```

```python
add_csv_data(df, SQLITE_PATH)
```

```python
def execute_query(db, query):
    conn = sqlite3.connect(db)
    cur = conn.cursor()
    cur.execute(query)
    rows = cur.fetchall()
    conn.close()
    return rows
```

```python
rows = execute_query(SQLITE_PATH, "select * from moves")
```
