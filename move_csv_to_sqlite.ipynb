```python
import pandas as pd
import sqlite3
```

```python
def create_database(csv_path, database_path, chunk_max):
    conn = sqlite3.connect(database_path)
    try:
        current = 0
        for chunk in pd.read_csv(csv_path, chunksize=1000000):
            print("Chunk: {}".format(current))
            chunk.drop(columns=chunk.columns[0], axis=1, inplace=True)
            chunk.to_sql("moves", conn, if_exists='append', index=False)
            current += 1
            if current == chunk_max:
                break
    finally:
        conn.close()
```

```python
CSV_PATH = './data/lichess_db_standard_rated_2019-04.csv'
SQLITE_PATH = './data/db1.sqlite3'
```

```python
create_database(CSV_PATH, SQLITE_PATH, None)
```

```python
SQLITE_2_PATH = './data/db2.sqlite3'
```

```python
create_database(CSV_PATH, SQLITE_2_PATH, 1)
```
